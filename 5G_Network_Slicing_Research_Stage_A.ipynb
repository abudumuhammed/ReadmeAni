```python
%%writefile /content/5G_Network_Slicing/ns-3-dev/scratch/no-slicing-5g-baseline.cc

#include "ns3/antenna-module.h"
#include "ns3/applications-module.h"
#include "ns3/core-module.h"
#include "ns3/flow-monitor-module.h"
#include "ns3/internet-module.h"
#include "ns3/mobility-module.h"
#include "ns3/nr-module.h"
#include "ns3/point-to-point-module.h"

using namespace ns3;

NS_LOG_COMPONENT_DEFINE("NoSlicing5GBaseline");

int
main(int argc, char* argv[])
{
    // ============================================================
    // 1. SIMULATION PARAMETERS
    // ============================================================

    uint16_t gNbNum = 1;

    // Total UEs:
    // 4 eMBB + 4 URLLC + 12 mMTC = 20 UEs
    uint16_t eMbbUeNum = 4;
    uint16_t urllcUeNum = 4;
    uint16_t mmtcUeNum = 12;

    uint16_t totalUeNum = eMbbUeNum + urllcUeNum + mmtcUeNum;

    double centralFrequency = 3.5e9;
    double bandwidth = 20e6;
    double totalTxPower = 43.0;

    uint16_t numerology = 1;

    Time simTime = Seconds(10.0);
    Time appStartTime = Seconds(1.0);

    std::string outputFile =
        "/content/5G_Network_Slicing/results/no_slicing_baseline.csv";

    CommandLine cmd;

    cmd.AddValue("simTime",
                 "Simulation duration",
                 simTime);

    cmd.AddValue("centralFrequency",
                 "NR carrier frequency",
                 centralFrequency);

    cmd.AddValue("bandwidth",
                 "NR bandwidth",
                 bandwidth);

    cmd.AddValue("totalTxPower",
                 "gNB total transmit power",
                 totalTxPower);

    cmd.AddValue("numerology",
                 "NR numerology",
                 numerology);

    cmd.Parse(argc, argv);

    // ============================================================
    // 2. CREATE OUTPUT DIRECTORY
    // ============================================================

    system("mkdir -p /content/5G_Network_Slicing/results");

    // ============================================================
    // 3. CREATE gNB AND UE NODES
    // ============================================================

    NodeContainer gNbNodes;
    gNbNodes.Create(gNbNum);

    NodeContainer eMbbUes;
    eMbbUes.Create(eMbbUeNum);

    NodeContainer urllcUes;
    urllcUes.Create(urllcUeNum);

    NodeContainer mmtcUes;
    mmtcUes.Create(mmtcUeNum);

    // ============================================================
    // 4. MOBILITY / POSITIONING
    // ============================================================

    MobilityHelper gnbMobility;
    Ptr<ListPositionAllocator> gnbPosition =
        CreateObject<ListPositionAllocator>();

    gnbPosition->Add(Vector(0.0, 0.0, 10.0));

    gnbMobility.SetPositionAllocator(gnbPosition);
    gnbMobility.SetMobilityModel("ns3::ConstantPositionMobilityModel");
    gnbMobility.Install(gNbNodes);

    MobilityHelper ueMobility;

    Ptr<ListPositionAllocator> uePosition =
        CreateObject<ListPositionAllocator>();

    // eMBB UEs
    for (uint16_t i = 0; i < eMbbUeNum; ++i)
    {
        uePosition->Add(
            Vector(10.0 + i * 5.0, 5.0, 1.5));
    }

    // URLLC UEs
    for (uint16_t i = 0; i < urllcUeNum; ++i)
    {
        uePosition->Add(
            Vector(10.0 + i * 5.0, -5.0, 1.5));
    }

    // mMTC UEs
    for (uint16_t i = 0; i < mmtcUeNum; ++i)
    {
        uePosition->Add(
            Vector(15.0 + i * 3.0, 10.0, 1.5));
    }

    ueMobility.SetPositionAllocator(uePosition);
    ueMobility.SetMobilityModel(
        "ns3::ConstantPositionMobilityModel");

    ueMobility.Install(eMbbUes);
    ueMobility.Install(urllcUes);
    ueMobility.Install(mmtcUes);

    // ============================================================
    // 5. NR / EPC HELPERS
    // ============================================================

    Ptr<NrPointToPointEpcHelper> nrEpcHelper =
        CreateObject<NrPointToPointEpcHelper>();

    Ptr<IdealBeamformingHelper> beamformingHelper =
        CreateObject<IdealBeamformingHelper>();

    Ptr<NrHelper> nrHelper =
        CreateObject<NrHelper>();

    nrHelper->SetEpcHelper(nrEpcHelper);
    nrHelper->SetBeamformingHelper(beamformingHelper);

    // ============================================================
    // 6. COMMON UNSLICED NR RESOURCE CONFIGURATION
    // ============================================================

    CcBwpCreator ccBwpCreator;

    const uint8_t numCc = 1;

    CcBwpCreator::SimpleOperationBandConf bandConf(
        centralFrequency,
        bandwidth,
        numCc);

    OperationBandInfo band =
        ccBwpCreator.CreateOperationBandContiguousCc(bandConf);

    nrHelper->SetGnbPhyAttribute(
        "Numerology",
        UintegerValue(numerology));

    nrHelper->SetGnbPhyAttribute(
        "TxPower",
        DoubleValue(totalTxPower));

    // ============================================================
    // 7. ANTENNAS
    // ============================================================

    nrHelper->SetGnbAntennaAttribute(
        "NumRows",
        UintegerValue(2));

    nrHelper->SetGnbAntennaAttribute(
        "NumColumns",
        UintegerValue(2));

    nrHelper->SetUeAntennaAttribute(
        "NumRows",
        UintegerValue(1));

    nrHelper->SetUeAntennaAttribute(
        "NumColumns",
        UintegerValue(1));

    // ============================================================
    // 8. INSTALL NR DEVICES
    // ============================================================

    NetDeviceContainer gNbDevices =
        nrHelper->InstallGnbDevice(
            gNbNodes,
            ccBwpCreator.GetAllBwps(band));

    NetDeviceContainer eMbbDevices =
        nrHelper->InstallUeDevice(
            eMbbUes,
            ccBwpCreator.GetAllBwps(band));

    NetDeviceContainer urllcDevices =
        nrHelper->InstallUeDevice(
            urllcUes,
            ccBwpCreator.GetAllBwps(band));

    NetDeviceContainer mmtcDevices =
        nrHelper->InstallUeDevice(
            mmtcUes,
            ccBwpCreator.GetAllBwps(band));

    // ============================================================
    // 9. INTERNET STACK
    // ============================================================

    InternetStackHelper internet;
    internet.Install(eMbbUes);
    internet.Install(urllcUes);
    internet.Install(mmtcUes);

    // ============================================================
    // 10. ASSOCIATE UEs WITH THE gNB
    // ============================================================

    nrHelper->AttachToClosestEnb(eMbbDevices, gNbDevices);
    nrHelper->AttachToClosestEnb(urllcDevices, gNbDevices);
    nrHelper->AttachToClosestEnb(mmtcDevices, gNbDevices);

    // ============================================================
    // 11. APPLICATIONS
    // ============================================================

    uint16_t dlPort = 5000;

    // -------------------------
    // eMBB: high-rate traffic
    // -------------------------

    for (uint16_t i = 0; i < eMbbUeNum; ++i)
    {
        Ptr<Node> ue = eMbbUes.Get(i);

        PacketSinkHelper sink(
            "ns3::UdpSocketFactory",
            InetSocketAddress(Ipv4Address::GetAny(),
                              dlPort));

        ApplicationContainer sinkApp =
            sink.Install(ue);

        sinkApp.Start(Seconds(0.5));
        sinkApp.Stop(simTime);

        OnOffHelper onoff(
            "ns3::UdpSocketFactory",
            InetSocketAddress(
                ue->GetObject<Ipv4>()->GetAddress(1, 0).GetLocal(),
                dlPort));

        onoff.SetAttribute(
            "DataRate",
            DataRateValue(DataRate("100Mbps")));

        onoff.SetAttribute(
            "PacketSize",
            UintegerValue(1200));

        onoff.SetAttribute(
            "StartTime",
            TimeValue(appStartTime));

        onoff.SetAttribute(
            "StopTime",
            TimeValue(simTime));

        dlPort++;
    }

    // -------------------------
    // URLLC: low-rate packets
    // -------------------------

    for (uint16_t i = 0; i < urllcUeNum; ++i)
    {
        Ptr<Node> ue = urllcUes.Get(i);

        PacketSinkHelper sink(
            "ns3::UdpSocketFactory",
            InetSocketAddress(Ipv4Address::GetAny(),
                              dlPort));

        ApplicationContainer sinkApp =
            sink.Install(ue);

        sinkApp.Start(Seconds(0.5));
        sinkApp.Stop(simTime);

        UdpClientHelper client(
            ue->GetObject<Ipv4>()->GetAddress(1, 0).GetLocal(),
            dlPort);

        client.SetAttribute(
            "MaxPackets",
            UintegerValue(100000));

        client.SetAttribute(
            "Interval",
            TimeValue(MilliSeconds(10)));

        client.SetAttribute(
            "PacketSize",
            UintegerValue(200));

        ApplicationContainer clientApp =
            client.Install(gNbNodes.Get(0));

        clientApp.Start(appStartTime);
        clientApp.Stop(simTime);

        dlPort++;
    }

    // -------------------------
    // mMTC: many low-rate devices
    // -------------------------

    for (uint16_t i = 0; i < mmtcUeNum; ++i)
    {
        Ptr<Node> ue = mmtcUes.Get(i);

        PacketSinkHelper sink(
            "ns3::UdpSocketFactory",
            InetSocketAddress(Ipv4Address::GetAny(),
                              dlPort));

        ApplicationContainer sinkApp =
            sink.Install(ue);

        sinkApp.Start(Seconds(0.5));
        sinkApp.Stop(simTime);

        UdpClientHelper client(
            ue->GetObject<Ipv4>()->GetAddress(1, 0).GetLocal(),
            dlPort);

        client.SetAttribute(
            "MaxPackets",
            UintegerValue(100000));

        client.SetAttribute(
            "Interval",
            TimeValue(MilliSeconds(100)));

        client.SetAttribute(
            "PacketSize",
            UintegerValue(100));

        ApplicationContainer clientApp =
            client.Install(gNbNodes.Get(0));

        clientApp.Start(appStartTime);
        clientApp.Stop(simTime);

        dlPort++;
    }

    // ============================================================
    // 12. FLOW MONITOR
    // ============================================================

    FlowMonitorHelper flowmonHelper;

    Ptr<FlowMonitor> monitor =
        flowmonHelper.InstallAll();

    // ============================================================
    // 13. SIMULATION
    // ============================================================

    Simulator::Stop(simTime);

    Simulator::Run();

    // ============================================================
    // 14. SAVE FLOW STATISTICS
    // ============================================================

    monitor->CheckForLostPackets();

    Ptr<Ipv4FlowClassifier> classifier =
        DynamicCast<Ipv4FlowClassifier>(
            flowmonHelper.GetClassifier());

    std::map<FlowId, FlowMonitor::FlowStats> stats =
        monitor->GetFlowStats();

    std::ofstream out(outputFile);

    out << "FlowId,Source,Destination,TxPackets,RxPackets,"
           "LostPackets,ThroughputMbps,MeanDelayMs,MeanJitterMs\n";

    for (const auto& flow : stats)
    {
        Ipv4FlowClassifier::FiveTuple t =
            classifier->FindFlow(flow.first);

        const FlowMonitor::FlowStats& st = flow.second;

        double duration =
            (st.timeLastRxPacket -
             st.timeFirstTxPacket).GetSeconds();

        double throughput =
            duration > 0
                ? (st.rxBytes * 8.0 /
                   duration / 1e6)
                : 0.0;

        double meanDelay =
            st.rxPackets > 0
                ? st.delaySum.GetMilliSeconds() /
                  st.rxPackets
                : 0.0;

        double meanJitter =
            st.rxPackets > 1
                ? st.jitterSum.GetMilliSeconds() /
                  (st.rxPackets - 1)
                : 0.0;

        out << flow.first << ","
            << t.sourceAddress << ","
            << t.destinationAddress << ","
            << st.txPackets << ","
            << st.rxPackets << ","
            << st.lostPackets << ","
            << throughput << ","
            << meanDelay << ","
            << meanJitter << "\n";
    }

    out.close();

    monitor->SerializeToXmlFile(
        "/content/5G_Network_Slicing/results/"
        "no_slicing_baseline_flowmon.xml",
        true,
        true);

    Simulator::Destroy();

    std::cout << "\n============================================\n";
    std::cout << "NO-SLICING 5G NR BASELINE COMPLETED\n";
    std::cout << "gNBs: " << gNbNum << "\n";
    std::cout << "eMBB UEs: " << eMbbUeNum << "\n";
    std::cout << "URLLC UEs: " << urllcUeNum << "\n";
    std::cout << "mMTC UEs: " << mmtcUeNum << "\n";
    std::cout << "Total UEs: " << totalUeNum << "\n";
    std::cout << "============================================\n";

    return 0;
}
```

```python
%cd /content/5G_Network_Slicing/ns-3-dev

!python3 - <<'PY'
from pathlib import Path

p = Path("scratch/no-slicing-5g-baseline.cc")
s = p.read_text()

# ------------------------------------------------------------
# Fix 1: Create the operation-band vector required by v5.1
# ------------------------------------------------------------

s = s.replace(
'''    OperationBandInfo band =
        ccBwpCreator.CreateOperationBandContiguousCc(bandConf);
''',
'''    OperationBandInfo band =
        ccBwpCreator.CreateOperationBandContiguousCc(bandConf);

    std::vector<std::reference_wrapper<OperationBandInfo>> bands;
    bands.emplace_back(band);

    auto allBwps = ccBwpCreator.GetAllBwps(bands);
'''
)

# Replace all incorrect GetAllBwps(band) calls
s = s.replace(
    "ccBwpCreator.GetAllBwps(band)",
    "allBwps"
)

# ------------------------------------------------------------
# Fix 2: AttachToClosestEnb -> AttachToClosestGnb
# ------------------------------------------------------------

s = s.replace(
    "AttachToClosestEnb",
    "AttachToClosestGnb"
)

p.write_text(s)

print("Baseline source updated successfully.")
print("Applied:")
print("  1. GetAllBwps() v5.1 API correction")
print("  2. AttachToClosestGnb() v5.1 API correction")
```
